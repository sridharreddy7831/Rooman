```python
# adding labels
plt.plot(a,b,marker='o',ms=15,mec='r',mfc='y')  
plt.xlabel('Income')
plt.ylabel('Outcome')
```

```python
# adding tittle
plt.plot(a,b,marker='o',ms=15,mec='r',mfc='y')  
plt.xlabel('Income')
plt.ylabel('Outcome')
plt.title('Line-Plot')
```

```python
# setting the title at left or right
plt.plot(a,b,marker='o',ms=15,mec='r',mfc='y')   
plt.xlabel('Income')
plt.ylabel('Outcome')
plt.title('Line-Plot',loc='left')
```

```python
plt.plot(a,b,marker='o',ms=15,mec='r',mfc='y')   
plt.xlabel('Income')
plt.ylabel('Outcome')
plt.title('Line-Plot',loc='right')
```

```python
# adding grids
plt.plot(a,b,marker='o',ms=15,mec='r',mfc='y')   
plt.xlabel('Income')
plt.ylabel('Outcome')
plt.title('Line-Plot',loc='right')
plt.grid()
```

```python
# adding grids for x-axis 
plt.plot(a,b,marker='o',ms=15,mec='r',mfc='y')   
plt.xlabel('Income')
plt.ylabel('Outcome')
plt.title('Line-Plot',loc='right')
plt.grid(axis='x')
```

```python
#adding grids for y-axis
plt.plot(a,b,marker='o',ms=15,mec='r',mfc='y')   
plt.xlabel('Income')
plt.ylabel('Outcome')
plt.title('Line-Plot',loc='right')
plt.grid(axis = 'y')
```

```python
# creating tittle for the subplots
plt.subplot(1,2,1)
plt.plot(a,b)
plt.title('Left-Plot')

plt.subplot(1,2,2)
plt.plot(c,d)
plt.title('Right-Plot')

plt.suptitle('Column-Wise')   #Tittle for entire plot
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('DataSets\diabetes.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
plt.plot(df['Pregnancies'],df['Glucose'],marker='o',ms=15,mec='r',mfc='y')
```

```python
plt.hist(df['Age']) # creating the histogram for age
```

```python
plt.pie(df['BMI'].head(10),labels=df['BMI'].head(10)) # creating pie chart for BMI of first 10 records
```

```python
plt.pie(df['BMI'].head(49),labels=df['BMI'].head(49),shadow=True) # creating pie chart for BMI of last 10 records
```

```python
plt.bar(df['Age'],df['Glucose']) # creating bar plot for pregnancies and glucose
```

```python
df
```

```python
plt.scatter(df['Age'].head(500),df['BloodPressure'].head(500)) # creating scatter plot for age and blood pressure of first 500 records
plt.colorbar() # creating color bar for the scatter plot
```

```python
# creating scatter plot for age and blood pressure of first 500 records
plt.scatter(df['Age'].head(500),df['BloodPressure'].head(500)) # creating scatter plot for age and blood pressure of first 500 records
plt.colorbar() # creating color bar for the scatter plot
plt.xlabel('Age')
plt.ylabel('Blood Pressure')
plt.title('Age vs Blood Pressure')
plt.show() # showing the plot
```

```python
# creating new dataframe with unique values of Age and SkinThickness
# and resetting the index
a = pd.DataFrame(df[['Age', 'SkinThickness']].drop_duplicates().reset_index(drop=True))
a
```

```python
plt.bar(a['Age'],a['SkinThickness']) # creating bar plot for age and skin thickness
plt.xlabel('Age')
plt.ylabel('Skin Thickness')
plt.title('Age vs Skin Thickness')
plt.show() # showing the plot
```
