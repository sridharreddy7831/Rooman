```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv('DataSets\clv_data.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['age'] = df['age'].fillna(df['age'].mean())
```

```python
df['days_on_platform'] = df['days_on_platform'].fillna(df['days_on_platform'].mean())
```

```python
df.isnull().sum()
```

```python
# IQR calculation
Q1 = df['days_on_platform'].quantile(0.25)
Q3 = df['days_on_platform'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find outliers
outliers = df[(df['days_on_platform'] < lower_bound) | (df['days_on_platform'] > upper_bound)]
num_outliers = outliers.shape[0]

print("Number of outliers for days_on_platform:", num_outliers)
```

```python
plt.boxplot(df['days_on_platform'])
```

```python
# IQR calculation
Q1 = df['income'].quantile(0.25)
Q3 = df['income'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find outliers
outliers = df[(df['income'] < lower_bound) | (df['income'] > upper_bound)]
num_outliers = outliers.shape[0]

print("Number of outliers for income:", num_outliers)
```

```python
plt.boxplot(df['income'])
```

```python
# IQR calculation
Q1 = df['purchases'].quantile(0.25)
Q3 = df['purchases'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find outliers
outliers = df[(df['purchases'] < lower_bound) | (df['purchases'] > upper_bound)]
num_outliers = outliers.shape[0]

print("Number of outliers for purchases:", num_outliers)
```

```python
# IQR calculation
Q1 = df['days_on_platform'].quantile(0.25)
Q3 = df['days_on_platform'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find outliers
outliers = df[(df['days_on_platform'] < lower_bound) | (df['days_on_platform'] > upper_bound)]
num_outliers = outliers.shape[0]

print("Number of outliers for days_on_platform:", num_outliers)

# Now fix the outliers (don't delete)
median_value = df['days_on_platform'].median()

# Replace only the outlier values with median
df.loc[(df['days_on_platform'] < lower_bound) | (df['days_on_platform'] > upper_bound), 'days_on_platform'] = median_value

# Done!
```

```python
# Check if any outliers remain
outliers_after = df[(df['days_on_platform'] < lower_bound) | (df['days_on_platform'] > upper_bound)]
print("Outliers remaining after replacement:", outliers_after.shape[0])
```

```python
import pandas as pd

# Assuming your DataFrame 'df' already exists

# Replace outliers in 'Income' with median (IQR method)
Q1 = df['income'].quantile(0.25)
Q3 = df['income'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
median_value = df['income'].median()

# Replace outliers
df['income'] = df['income'].apply(
    lambda x: median_value if (x < lower_bound or x > upper_bound) else x
)

print("Outliers have been replaced with the median successfully!")

outliers_after = df[(df['income'] < lower_bound) | (df['income'] > upper_bound)]
print("Outliers remaining:", outliers_after.shape[0])
```

### Scaling in Normalization and Standaization

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Assuming your original DataFrame is df

# Initialize the scalers
scaler_standard = StandardScaler()  # Z-Score Standardization
scaler_minmax = MinMaxScaler()  # Min-Max Normalization

# Fit and transform the 'income' column
income_std = scaler_standard.fit_transform(df[['income']])
income_Nor = scaler_minmax.fit_transform(df[['income']])

# Create a new DataFrame for scaled income
df_income = pd.DataFrame()
df_income['income_std'] = income_std.flatten()  # Flatten to ensure 1D
df_income['income_Nor'] = income_Nor.flatten()  # Flatten to ensure 1D

# Print the first 10 rows
print(df_income.head(10))

# Print the last 10 rows
print(df_income.tail(10))
```
