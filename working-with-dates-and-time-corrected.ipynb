```python
# Converting 'date' column to datetime type
date['date'] = pd.to_datetime(date['date'])
date.info()
```

```python
# Extracting year from the date
date['date_year'] = date['date'].dt.year
date.sample(5)
```

```python
# Extracting month (number)
date['date_month_no'] = date['date'].dt.month
date.head()
```

```python
# Extracting month (name)
date['date_month_name'] = date['date'].dt.month_name()
date.head()
```

```python
# Extracting day
date['date_day'] = date['date'].dt.day
date.head()
```

```python
# Day of week (as number)
date['date_dow'] = date['date'].dt.dayofweek
date.head()
```

```python
# Day of week (as name)
date['date_dow_name'] = date['date'].dt.day_name()
columns_to_drop = ['product_id', 'city_id', 'orders']
existing_cols = [col for col in columns_to_drop if col in date.columns]
date.drop(columns=existing_cols).head()
```

```python
# Is weekend?
date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1, 0)
columns_to_drop = ['product_id', 'city_id', 'orders']
existing_cols = [col for col in columns_to_drop if col in date.columns]
date.drop(columns=existing_cols).head()
```
