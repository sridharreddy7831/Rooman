```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('DataSets\diabetes.csv')
```

```python
df.head()
```

```python
#displot function is used to create a distribution plot
# here bins=7 means the data will be divided into 7 equal parts
# and kde=True means the kernel density estimation will be plotted

sns.displot(df['BMI'],bins=7,kde=True) 
```

```python
# this is only to show the kde plot and not the histogram
# here shade=True means the area under the curve will be shaded


sns.kdeplot(df['BMI'],shade=True)
```

```python
sns.lineplot(df['Age'][:20])
```

```python
# this will show the count of each value in the Outcome column
# here hue='Outcome' means the data will be divided into 2 parts based on the Outcome column
# and the count of each part will be shown

sns.countplot(x='Outcome',data=df)


```

```python
sns.boxplot(df['Glucose'])
```

```python
sns.violinplot(data=df,x='Outcome',y='Glucose')
```

```python
a = df.corr()
sns.heatmap(a,annot=True)
```

```python
sns.pairplot(df,hue='Outcome',diag_kind='kde')
```

```python
sns.pairplot(df,hue='Outcome',diag_kind='kde')
plt.tight_layout()
```

```python
df.corr()
```

```python
plt.tight_layout() #this will adjust the subplots to fit into the figure area.
```

```python
plt.figure(figsize=(10,7)) #this will set the figure size to 10x5 inches
sns.scatterplot(x='Age',y='Glucose',data=df,hue='Outcome')
```

```python
import plotly.express as px

px.scatter(df,x='Age',y='Glucose',color='Outcome',title='Age vs Glucose',size_max=10,template='plotly_dark')
```
