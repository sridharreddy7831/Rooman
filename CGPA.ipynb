```python
def calculate_cgpa(sgpas, total_credits):
    if len(sgpas) != len(total_credits):
        raise ValueError("Length of SGPAs and total credits must be the same")

    total_weighted_score = sum(s * c for s, c in zip(sgpas, total_credits))
    total_credits_sum = sum(total_credits)

    cgpa = total_weighted_score / total_credits_sum
    return round(cgpa, 2)


# Example usage
sgpas = [8.18, 8.18, 8.78, 8.74, 8.60, 8.19, 7.91, 10]  # Replace with your actual SGPAs
credits = [19.5, 19.5, 24.5, 21.5, 21.5, 21.5, 23, 12]    # Replace with total credits for each semester

cgpa = calculate_cgpa(sgpas, credits)
print("CGPA:", cgpa)

# To convert CGPA to percentage:
percentage = (cgpa - 0.5) * 10
print("Equivalent Percentage:", round(percentage, 2))
```
